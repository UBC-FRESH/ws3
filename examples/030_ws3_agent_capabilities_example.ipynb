# ws3 Agent Capabilities — IPython / Jupyter Demo

This notebook demonstrates the ws3 agent capabilities via IPython line magics.
A live `ForestModel` named `fm` is set up first, then the magics are used to:

- **`%ws3_hint`** — general modelling guidance with verifiable symbol & URL references
- **`%build_mask`** — build a development-type mask from a plain-English description
- **`%explain_exception`** — explain a Python exception in plain language
- **`%diagnose_import`** — diagnose a failing Woodstock section import
- **`%rtfm`** — route a goal or error to the correct capability

The `fm` ForestModel is discovered automatically — no explicit model argument is needed.

## Build a ForestModel

Set up a minimal model that will be in scope for the capabilities below.

```python
from pathlib import Path
from ws3.forest import ForestModel

MODEL_DIR = Path('/home/gep/projects/ws3/examples/data/woodstock_model_files_tsa24_clipped')
MODEL_NAME = 'tsa24_clipped'

fm = ForestModel(
    model_name=MODEL_NAME,
    model_path=str(MODEL_DIR),
    base_year=2020,
    horizon=10,
    period_length=10,
    max_age=1000,
)
fm.import_landscape_section()
fm.import_areas_section(convert_periods_to_years=10)
fm.import_actions_section(convert_periods_to_years=10)
fm.import_transitions_section(convert_periods_to_years=10)

print('Model loaded:', fm.model_name)
print('Actions:', sorted(fm.actions))
```

## %ws3_hint — general modelling guidance

Ask a "how do I..." question. The model returns suggested steps plus cited
ws3 symbols (validated against the installed package) and doc URLs (validated
with HTTP 200). Fabricated APIs are caught automatically.

```python
%ws3_hint How do I add a fire disturbance?
```

```python
%ws3_hint How do I add a partial cut harvest action to my ws3 model?
```

```python
%ws3_hint How do I run multi-period simulations with different management scenarios?
```

```python
%ws3_hint Can you help me write an "area control" harvesting heuristic to simulate maximum long-term sustainable yield?
```

## %build_mask — build a development-type mask

Describe what stands you want in plain English, using the model's theme names
and codes. Age is not a ws3 stratification theme, so age-based filtering belongs
in separate stand-level logic rather than in a development-type mask. The mask is
validated against the live ForestModel — if no real development types match, the
proposal is rejected with a clear error.

```python
%build_mask all pine stands
```

## %explain_exception — explain a Python exception

Paste an error message or traceback. The explanation cites only real ws3
symbols that actually exist in the installed package.

```python
%explain_exception KeyError: 'theme not found in fm.theme_names'
```

## %rtfm — route to the correct capability

Give a goal or error description and let the router pick the right capability
and parameters automatically.

```python
%rtfm How do I select all pine stands in my model?
```

```python
%rtfm I got a KeyError when trying to import my landscape file
```

## Under the hood — Python API

The magics are thin wrappers around the same capability objects you can call
directly from Python, for use in scripts or custom workflows.

```python
# Direct Python API — uses the same settings.json model config as the magics
from ws3.agent.ipython_magics import _make_config, _make_provider
from ws3.agent.capabilities.build_mask import BuildMask, MaskRequest

config = _make_config()
provider = _make_provider(config)

cap = BuildMask(fm)
result = cap.run(
    MaskRequest(description='all pine stands'),
    provider=provider,
    config=config,
    context=fm,
)

print('OK:', result.ok)
if result.errors:
    for err in result.errors:
        print('Error:', err)
else:
    print('Mask:', result.value.mask)
```
